```python
import chromadb

client = chromadb.PersistentClient(path="../data/chromadb")

collection = client.get_or_create_collection(name="my_docs")

print("ChromaDB ready!")
```

```python
sentences = [
    "Artificial Intelligence is transforming the world.",
    "Machine learning is a subset of AI.",
    "I love eating pizza.",
    "Deep learning uses neural networks.",
    "Natural language processing helps computers understand text."
]
 
collection.add(documents = sentences, ids=[f"doc_{i}" for i in range(len(sentences))]
)


print(f"Total stored: {collection.count()}")
```

```python
results = collection.query(
    query_texts=["What is machine learning?"],
    n_results=2
)

print("Top matches:")
for i, doc in enumerate(results['documents'][0]):
    print(f"\nRank {i+1}: {doc}")
    print(f"Distance: {results['distances'][0][i]:.4f}")
```

```python

collection2 = client.get_or_create_collection(name="my_docs_with_meta")

collection2.add(
    documents=sentences,
    ids=[f"doc_{i}" for i in range(len(sentences))],
    metadatas=[{"source": "textbook", "page": i+1} for i in range(len(sentences))]
)

results2 = collection2.query(
    query_texts=["What is machine learning?"],
    n_results=2
)

print("Top matches with metadata:")
for i, doc in enumerate(results2['documents'][0]):
    print(f"\nRank {i+1}: {doc}")
    print(f"Source: {results2['metadatas'][0][i]}")
```
