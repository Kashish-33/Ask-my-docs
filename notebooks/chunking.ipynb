```python
text = """
Artificial Intelligence is transforming the world. 
It is used in healthcare, finance, education, and many other fields.
Machine learning is a subset of AI that allows systems to learn from data.
Deep learning uses neural networks to solve complex problems.
Natural Language Processing helps computers understand human language.
Computer vision enables machines to interpret visual information.
AI is becoming increasingly important in our daily lives.
Researchers are constantly working on making AI more efficient and accurate.
Ethical considerations in AI are becoming more important than ever.
The future of AI holds immense possibilities for humanity.
"""

print(text)
```

import nltk
nltk.download('punkt_tab')

sentences = nltk.sent_tokenize(text)
print(f"Total sentences: {len(sentences)}")
print(sentences)

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size = 200,
    chunk_overlap = 50
)

chunks = splitter.split_text(text)

print(f"Total chunks: {len(chunks)}")
print("---")
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}: {chunk}")
    print("---")
```

```python
# chunk_overlap kyun rakha? Ye dekho:
# Last line of chunk 1 aur first line of chunk 2 overlap karti hai
# Isse context lost nahi hota chunks ke beech mein!

print("Chunk 1 end:", chunks[0][-50:])
print("Chunk 2 start:", chunks[1][:50])
```
