```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')

sentences = [
    "Artificial Intelligence is transforming the world.",
    "Machine learning is a subset of AI.",
    "I love eating pizza.",
]

embeddings = model.encode(sentences)

print(f"Total embeddings: {len(embeddings)}")
print(f"Each embedding size: {len(embeddings[0])}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Compare karo — AI wali sentences kitni similar hain?
sim_1_2 = cosine_similarity([embeddings[0]], [embeddings[1]])[0][0]
sim_1_3 = cosine_similarity([embeddings[0]], [embeddings[2]])[0][0]

print(f"AI sentence vs ML sentence: {sim_1_2:.4f}")
print(f"AI sentence vs Pizza sentence: {sim_1_3:.4f}")
```

```python
import faiss
import numpy as np 

embeddings_matrix = np.array(embeddings).astype('float32')

dimension = 384
index = faiss.IndexFlatL2(dimension)


index.add(embeddings_matrix)

print(f"Total vectors stored: {index.ntotal}")
```

```python
query = " What is machine learning?"

query_embedding = model.encode([query]).astype("float32")

k=2
distances, indices = index.search(query_embedding, k)

print(f"Query: {query}")
print(f"\nTop {k} matching sentences:")
for i, idx in enumerate(indices[0]):
    print(f"\nRank {i+1}: {sentences[idx]}")
    print(f"Distance: {distances[0][i]:.4f}")
```
